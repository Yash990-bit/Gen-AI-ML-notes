# Problem with Lasso

```
Lasso bolta hai:

"Main useless features ko 0 kar dunga."

Example

Feature	Coefficient
Study	8
Attendance	3
Sleep	2
Shoe Color	0
Hair Length	0

Ye achha hai.

Lekin agar Study Hours aur Practice Hours dono almost same information dete hain (highly correlated features), to Lasso kabhi-kabhi sirf ek ko rakhta hai aur dusre ko hata deta hai.

Kabhi Study rakhega.

Kabhi Practice rakhega.

Ye unstable ho sakta hai.

 # Solution = Elastic Net

Elastic Net bolta hai:

Main Ridge aur Lasso dono ki achhi baatein use karunga.

Ridge ki tarah coefficients ko shrink bhi karunga.

Lasso ki tarah useless features ko remove bhi karunga.

# Loss Function

Elastic Net ka loss function:

Loss = Error
     + λ₁(|w₁| + |w₂| + ...)
     + λ₂(w₁² + w₂² + ...)

Matlab

L1 Penalty (Lasso)

L2 Penalty (Ridge)

Dono ek saath.

# Alpha ka role

Elastic Net me Alpha (α) decide karta hai ki Lasso aur Ridge ka kitna mix hoga.

α = 0 → Pure Ridge

α = 1 → Pure Lasso

0 < α < 1 → Ridge + Lasso ka combination

Example:

α = 0.2 → 20% Lasso + 80% Ridge

α = 0.5 → 50% Lasso + 50% Ridge

α = 0.8 → 80% Lasso + 20% Ridge

 # Kab use karte hain?

Elastic Net tab best hota hai jab:

Dataset me bahut saare features hon.

Features ek dusre se correlated hon.

Kuch features ko remove bhi karna ho.

Aur baaki coefficients ko bhi control me rakhna ho.

# CODE

````python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression,Ridge,Lasso,ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
````

````python

X,y = load_diabetes(return_X_y=True)
````

````python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
````

````python
# Linear Regression
reg = LinearRegression()
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
````

````python
# Ridge
reg = Ridge(alpha=0.1)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
````

````python
# Lasso
reg = Lasso(alpha=0.01)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
````

````python
# ElasticNet
reg = ElasticNet(alpha=0.005,l1_ratio=0.9)
reg.fit(X_train,y_train)
y_pred = reg.predict(X_test)
r2_score(y_test,y_pred)
````

# Ridge/Lasso: alpha = Lambda (penalty kitni strong hogi).
# Elastic Net: alpha = L1 aur L2 ka mixing ratio, jabki regularization ki overall strength alag parameter (jaise lambda ya scikit-learn me alpha + l1_ratio) se control ki ja sakti hai.
