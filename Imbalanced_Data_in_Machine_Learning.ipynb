# Imbalanced Data

Jab dataset mein ek class ke examples bahut zyada ho aur doosri class ke bahut kam.

Real-Life Example

Socho ek 1000 students ka dataset hai:

✅ 950 students = Not Fraud

🚨 50 students = Fraud

Yaha:

Not Fraud = 950 (95%)

Fraud = 50 (5%)

Toh dataset imbalanced hai because classes ka distribution equal nahi hai.

# "Jab ek class ki population bahut zyada aur doosri class ki bahut kam ho → Imbalanced Data."

# ⚠️ Main Problem

Imbalanced data mein:

High Accuracy ≠ Good Model

Example:

| Actual     | Prediction   |
| ---------- | ------------ |
| 950 Normal | 950 Normal ✅ |
| 50 Fraud   | 50 Normal ❌  |


Accuracy = 95%

Lekin model ne ek bhi fraud detect nahi kiya! 🤦

Isliye imbalanced data mein hum Precision, Recall, F1-Score, Confusion Matrix jaise metrics ko bhi dekhte hain.

# Problems with Imbalanced Data

1. Model Majority Class ki taraf Bias ho jata hai
Model jis class ke examples zyada dekhta hai, usi ko baar-baar predict karne lagta hai.

Example:

950 Normal 🟢 + 50 Fraud 🔴

Model har case ko Normal bol de → 95% accuracy, but fraud detect hi nahi hua.

2. Accuracy Misleading ho sakti hai
Imbalanced dataset mein high accuracy ka matlab ye nahi ki model achha hai.

Example:

950 Normal + 50 Fraud

Agar model sabko Normal bole:

Accuracy = 95% ✅
Fraud Detection = 0% ❌

3. Minority Class Ignore ho sakti hai
Jo class kam hai, model usko properly learn nahi kar pata.

Example:

Fraud cases kam hain → model Fraud ko identify karne mein weak ho sakta hai.

4. Important Cases Miss ho sakte hain ⚠️
Minority class often important hoti hai.

Examples:

🏦 Fraud Detection → Fraud miss hona dangerous

🏥 Disease Detection → Disease patient miss hona dangerous

🔐 Cybersecurity → Attack miss hona dangerous

5. Poor Generalization

Model training data mein majority class ko toh achhe se predict karta hai, but new minority examples par fail ho sakta hai.

#  Why studying imbalanced data is important?

Imbalanced Data ko study karna important hai kyunki agar hum imbalance ko ignore kar dein, toh model misleadingly achha lag sakta hai but actually important cases mein fail ho sakta hai.

# it helps us build models that correctly identify minority classes, avoid misleading accuracy, choose appropriate evaluation metrics, and perform reliably in real-world applications.



# Undersampling

Majority class ke kuch data points ko hata dena, taaki classes ka balance improve ho.

Example

Suppose humare paas:

🟢 Normal = 1000

🔴 Fraud = 100

Dataset heavily imbalanced hai.

Undersampling mein hum Normal ke kuch samples remove kar dete hain:

🟢 Normal = 200

🔴 Fraud = 100

Ab imbalance kam ho gaya.

 Example

Socho:

🍕 100 Cheese pieces

🍄 10 Mushroom pieces

Cheese bahut zyada hai.

Hum 60 cheese pieces hata dete hain:

🍕 40 Cheese

🍄 10 Mushroom

➡️ Ye Undersampling hai.

🔄 How it works?

Imbalanced Data

⬇️

Majority Class identify karo

⬇️

Majority ke kuch samples remove karo


⬇️

Classes more balanced

⬇️

Model train karo

# Main Problem

Undersampling ka disadvantage:

Hum useful data bhi hata sakte hain.

Example:

1000 Normal → 200 Normal

800 data points remove ho gaye 😬

Isse model ke paas less information available ho sakti hai.

#  Oversampling
Minority class ke data ko increase karna, taaki dataset more balanced ho jaye.

Example

Suppose:

🟢 Normal = 1000

🔴 Fraud = 100

Yaha Fraud minority class hai.

Oversampling mein hum Fraud ke samples ko increase karte hain:

🟢 Normal = 1000

🔴 Fraud = 500

➡️ Ab classes pehle se more balanced hain.

Example

Socho:

🍕 100 Cheese pieces

🍄 10 Mushroom pieces

Mushroom bahut kam hai.

Hum mushroom samples ko duplicate/generate karke increase kar dete hain:

🍕 100 Cheese

🍄 50 Mushroom

➡️ Oversampling!

🔄 How it works?

Imbalanced Data

⬇️

Minority Class identify karo

⬇️

Minority samples increase karo

⬇️

Dataset more balanced

⬇️

Model train karo

# Common Oversampling Techniques

1. Random Oversampling

Minority samples ko randomly duplicate karte hain.

2. SMOTE

SMOTE sirf copy nahi karta; existing minority points ke beech new synthetic samples create karta hai.

# Problem

Agar hum same samples ko baar-baar copy karein:

100 Fraud → 1000 Fraud

Toh model un examples ko memorize kar sakta hai.

Isse Overfitting ho sakti hai. 😬

# SMOTE
Synthetic Minority Over-sampling Technique

use Imbalanced Data ko balance karne ke liye hota hai.

Ye minority class ke naye synthetic (artificial) data points create karta hai.

Example

Suppose:

🟢 Normal = 1000

🔴 Fraud = 100

Fraud minority class hai.

Random Oversampling mein hum same fraud points ko copy karte:

A → A → A → A

But SMOTE kya karta hai?

Do existing minority points ke beech ek naya point create karta hai:

A ● -------- ● B

SMOTE:

A ● ---- ● New ---- ● B

➡️ New point = Synthetic Data

# SMOTE Work ?

Step 1: Minority class identify karo

⬇️

Step 2: Ek minority data point choose karo

⬇️

Step 3: Uske nearest minority neighbors find karo

⬇️

Step 4: Unke beech new point create karo

⬇️

Step 5: Repeat until minority class sufficiently balanced ho jaye.

# Formula

SMOTE roughly:

New Point = Original Point + Random × (Neighbor − Original Point)

Example:

A = 10

B = 20

Random = 0.5

New point:

10 + 0.5 × (20 − 10)

= 15

➡️ 15 ek synthetic point hai.

# Ensemble Learning

Multiple ML models ko combine karke ek stronger/better model banana.

Example

Suppose tumhe decide karna hai:

“Ye movie achhi hai ya nahi?” 🎬

Tum ek friend se nahi poochte.

Tum 5 friends se poochte ho:

👨 Friend 1 → 👍

👩 Friend 2 → 👍

👨 Friend 3 → 👎

👩 Friend 4 → 👍

👨 Friend 5 → 👍

4 friends = Good

1 friend = Bad

➡️ Majority ke basis par decision: Movie Good! ✅

Exactly yahi idea Ensemble Learning mein hai.

# ML mein kaise?

Suppose humare paas 3 models hain:

🌳 Decision Tree → Fraud

🌲 Random Forest → Fraud

📈 Logistic Regression → Not Fraud

Majority:

2 → Fraud

1 → Not Fraud

➡️ Final Prediction = Fraud 🚨

Multiple models ki predictions combine karke final prediction banaya.

# Why Ensemble Learning?

Single model kabhi-kabhi:

❌ Overfit kar sakta hai

❌ Underfit kar sakta hai

❌ Wrong prediction de sakta hai

But multiple models ko combine karne se:

✅ Accuracy improve ho sakti hai

✅ Stability improve hoti hai

✅ Generalization better ho sakti hai

✅ Individual model ki weaknesses reduce ho sakti hain

# Cost Sensitive Learning


Model ko different mistakes ki different “cost” dena.

Har mistake equally dangerous nahi hoti.

Model ko sikhate hain ki important mistake ko zyada seriously lena hai.

# Example: Fraud Detection

Suppose bank transaction ko predict karna hai:

🟢 Normal
🔴 Fraud

Do mistakes possible hain:

Case 1:

Actual = Fraud 🔴

Model = Normal 🟢

➡️ Bank ne fraud ko miss kar diya 😱

Case 2:

Actual = Normal 🟢

Model = Fraud 🔴

➡️ Normal transaction ko fraud bol diya.

Dono mistakes hain, but Case 1 zyada costly ho sakta hai.

So hum model ko bolenge:

🚨 Fraud miss karne ki cost > Normal ko fraud bolne ki cost

# Imbalanced Data mein kyun useful?

Imbalanced dataset mein minority class often important hoti hai.

Example:

990 Normal 🟢

10 Fraud 🔴

Normal examples bahut zyada hain.

Agar model majority class ko blindly follow karega, toh fraud miss kar sakta hai.

Cost-Sensitive Learning minority class ki mistakes ko higher cost/weight dekar model ko un cases par more focus karwa sakta hai.

# Other Techniques for Imbalanced Data

1. Threshold Moving 🎚️

Normally model ka threshold 0.5 hota hai.
Hum threshold ko change karke minority class ko detect karna easier bana sakte hain.

Threshold change karo → prediction decision change karo.

2. Class Weighting ⚖️\
Minority class ko higher weight dete hain.

Example:

Normal → Weight = 1

Fraud  → Weight = 10

Model fraud ki mistakes ko zyada seriously lega.

Important class = Higher Weight

3. Balanced Ensemble

Multiple models train karte hain while ensuring each model gets a more balanced dataset.

Examples:

Balanced Random Forest

EasyEnsemble

Multiple models + Balanced data = Stronger model

4. Data Augmentation
Minority class ke new examples create karte hain.

SMOTE iska popular example hai.

New minority examples → Better learning
