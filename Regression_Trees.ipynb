# We need a Regression Tree to predict continuous numerical values when the relationship between input features and the target is complex or non-linear. It splits the data into smaller groups using decision rules and predicts the average value of each group, making it accurate, flexible, and easy to interpret

# Residual Error

Residual Error = Actual Value − Predicted Value

Residual = 0 → Perfect prediction ✅

Residual bada → Prediction kharab ❌

# Bias-Variance Tradeoff Kya Hai?

Tradeoff ka matlab hai balance banana.

Agar tum Bias kam karoge...

↓

Model complex hoga.

↓

Variance badh jayegi.

Agar Variance kam karoge...

↓

Model simple hoga.

↓

Bias badh jayega.

Dono ko ek saath minimum rakhna mushkil hota hai.

Isi balance ko Bias-Variance Tradeoff kehte hain.

```
| Bias                  | Variance                         |
| --------------------- | -------------------------------- |
| Wrong assumptions     | Prediction bahut change hoti hai |
| Model simple hota hai | Model complex hota hai           |
| Underfitting          | Overfitting                      |
| High training error   | Low training error               |
| High testing error    | High testing error               |
```

# Bias-Variance Tradeoff is the process of finding the right balance between bias and variance so that the model neither underfits nor overfits and performs well on unseen data.


# Finding the splitting criteria

Splitting Criteria Kya Hai?

Regression Tree me har node par bahut saare possible questions ho sakte hain.

Jaise:

House Size > 700?

House Size > 900?

House Size > 1200?

Ab Tree ko decide karna hai ki kaunsa question puche?

👉 Isi decision ko Splitting Criteria kehte hain.

Splitting Criteria ek rule hai jo decide karta hai ki data ko kis point par split karna hai taaki prediction sabse accurate ho.

# Regression Tree ka Goal

Regression Tree chahta hai ki:

Ek group ke andar sab target values ek jaisi (similar) ho.

Har group ka average prediction actual values ke bahut close ho.

Example:

Group 1

25

30

35

Ye values ek dusre ke paas hain. 👍

Group 2

25

80

35

Ye values bahut alag-alag hain. 👎

Regression Tree hamesha Group 1 jaisa group banana chahta hai.

# Similarity Kaise Measure Karte Hain?

Regression Tree generally Variance ya Mean Squared Error (MSE) use karta hai.

Simple language me:

Dekho group ke numbers average se kitne door hain.

Agar sab average ke paas hain → Good Split ✅

Agar bahut door hain → Bad Split ❌

# Tree Kaise Decide Karta Hai?

Tree har possible split ko try karta hai.

For example:

Size > 650

Size > 750

Size > 900

Size > 1000

Size > 1300

Har split ke baad:

Left group ka error nikalta hai.

Right group ka error nikalta hai.

Dono ko add karta hai.

Jis split ka total error sabse kam hota hai, wahi choose kar leta hai.

# 10. Univariate Regression

````python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor
from dtreeviz import dtreeviz # Explicitly import dtreeviz function (already imported globally, but good practice)

df_cars = pd.read_csv("cars.csv")
X, y = df_cars[['WGT']], df_cars['MPG']

dt = DecisionTreeRegressor(max_depth=3, criterion="absolute_error")
dt.fit(X, y)

# Use the main dtreeviz function for univariate regression visualization
viz = dtreeviz(dt,
               X,
               y,
               feature_names=["WGT"],
               target_name="MPG",
               title="Univariate Regression Tree (Weight vs MPG)")
viz # To display the visualization
````
