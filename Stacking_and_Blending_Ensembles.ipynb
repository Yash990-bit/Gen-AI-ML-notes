# Stacking

Imagine a Hospital 🏥

A patient is sick.

Instead of asking one doctor, you ask three specialists.

👨‍⚕️ Doctor 1 → Heart specialist

👨‍⚕️ Doctor 2 → Lung specialist

👨‍⚕️ Doctor 3 → Brain specialist

Each doctor gives their opinion.

Doctor	      Prediction

Heart	        Disease

Lung	        Healthy

Brain	        Disease

Now...

Instead of deciding yourself...

You send these opinions to a Senior Doctor.

The senior doctor doesn't look at the patient.

He only looks at the three doctors' opinions and makes the final decision.

This is Stacking.

# In Machine Learning

# Difference Between Bagging, Boosting, and Stacking

| Bagging                         | Boosting                                                       | Stacking                                                                 |
| ------------------------------- | -------------------------------------------------------------- | ------------------------------------------------------------------------ |
| Many models work independently. | Models learn one after another, each fixing previous mistakes. | Many different models make predictions, and another model combines them. |
| Final answer = Voting/Average.  | Final answer = Weighted combination.                           | Final answer = Meta-model prediction.                                    |

# Blending vs Stacking

| Blending                        | Stacking                                  |
| ------------------------------- | ----------------------------------------- |
| Validation set use karta hai    | K-Fold Cross Validation use karta hai     |
| Simple hai                      | Thoda complex hai                         |
| Fast hai                        | Slow hai                                  |
| Less computation                | More computation                          |
| Validation data kam ho jata hai | Pure training data ka better use hota hai |

# Why Stacking Uses K-Fold?

# Why Use Multiple Layers?

Kabhi-kabhi ek Meta Model bhi sab patterns nahi samajh pata.

To uske upar ek aur Meta Model laga dete hain.

Har layer previous layer ki predictions se aur achha decision lena seekhti hai.

# Multilayer Stacking is an extension of stacking where instead of using a single meta-model, multiple layers of meta-models are used. The predictions of base models are passed to first-level meta-models, whose outputs are then used by another higher-level meta-model to produce the final prediction. It can capture more complex relationships but increases computation, training time, and the risk of overfitting.

# Blending is an ensemble learning technique in which multiple base models are trained on the training data, their predictions are generated on a separate hold-out validation set, and these predictions are used to train a meta-model that makes the final prediction.

Blending ek ensemble learning technique hai jisme multiple base models (jaise Random Forest, XGBoost, SVM) ko training data par train kiya jata hai. Phir ye models hold-out validation set par predictions dete hain. In predictions ko meta-model train karne ke liye use kiya jata hai, jo final prediction karta hai.
