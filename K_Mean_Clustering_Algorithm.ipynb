```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

data = {
    'Age': [22,25,24,45,46,48,30,32,31],
    'Income': [25000,27000,26000,80000,82000,85000,45000,47000,46000]
}

df = pd.DataFrame(data)

# Scatter Plot
plt.scatter(df['Age'], df['Income'])
plt.xlabel("Age")
plt.ylabel("Income")
plt.show()

# Model
kmeans = KMeans(n_clusters=3, random_state=42)

# Train
kmeans.fit(df)

# Predict
df['Cluster'] = kmeans.predict(df)

print(df)

# Centroids
print(kmeans.cluster_centers_)

# Plot Clusters
plt.scatter(df['Age'], df['Income'], c=df['Cluster'])

plt.scatter(
    kmeans.cluster_centers_[:,0],
    kmeans.cluster_centers_[:,1],
    color='red',
    marker='X',
    s=200
)

plt.xlabel("Age")
plt.ylabel("Income")
plt.show()
```

# How Distance is Calculated?

K-Means usually Euclidean Distance use karta hai.

Formula

Distance = √[(x₂-x₁)² + (y₂-y₁)²]

Example

Point

A = (2,3)

Centroid = (5,7)

Distance

√[(5-2)²+(7-3)²]

= √(9+16)

= √25

=5

Jo centroid sabse paas hoga usi cluster me point chala jayega.

# What is Inertia?

Ye bahut popular interview question hai.

Inertia = Cluster ke andar points kitne close hain.

Formula (conceptually)

Inertia =
Σ (Distance from Point to Centroid)²
Small Inertia = Better Clustering ✅
Large Inertia = Poor Clustering ❌

Python

print(kmeans.inertia_)

# Initialization

K-Means pehle centroids randomly choose karta hai.

Kabhi random choice kharab ho sakti hai.

Solution

K-Means++

Ye smart tareeke se initial centroids choose karta hai.

Python me default

KMeans(init="k-means++")

# Parameters
KMeans(
    n_clusters=3,
    init="k-means++",
    max_iter=300,
    random_state=42
)

| Parameter    | Meaning                         |
| ------------ | ------------------------------- |
| n_clusters   | Number of clusters              |
| init         | Centroid initialize kaise honge |
| max_iter     | Maximum iterations              |
| random_state | Same result ke liye             |
