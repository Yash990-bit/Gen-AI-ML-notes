"Ek intelligent model se better, kai average models milkar zyada accurate decision le sakte hain."

# Example

Maan lo exam me ek difficult question hai.

Agar sirf 1 student se answer pucho, to wo galat bhi ho sakta hai.

Agar 10 students se answer pucho aur majority ka answer lo, to sahi answer milne ke chances zyada hain.

Yahi Ensemble Learning hai.

1 Student = 1 Machine Learning Model

10 Students = Multiple Models

Majority Vote = Final Prediction

# Core behind ensemble

Train Multiple Models

          ↓
Har model apni prediction deta hai

          ↓
Sab predictions ko combine karo

          ↓
Final prediction zyada accurate aur robust hoti hai

# Ensemble Learning ek technique hai jisme multiple machine learning models ko combine karke ek final prediction banayi jaati hai jo aksar ek single model se better hoti hai.

# Type of ensemble learning

# 1. Bagging (Bootstrap Aggregating)

👉 Sab models ek saath (parallel) train hote hain.

Har model alag-alag random data dekhta hai.

Phir sabki prediction combine ki jaati hai.

# Workflow

```
Dataset
   │
   ├── Random Sample 1 → Model 1
   ├── Random Sample 2 → Model 2
   ├── Random Sample 3 → Model 3
   └── Random Sample 4 → Model 4

            ↓
     Combine Predictions
            ↓
      Final Prediction
```


# Goal

Overfitting kam karna

Accuracy improve karna

# Example
✅ Random Forest

# 2. Boosting

👉 Models ek ke baad ek (sequentially) train hote hain.

Har naya model pichle model ki mistakes ko improve karta hai.

# Workflow

```
Dataset
   │
Model 1
   │
Mistakes
   │
Model 2
   │
Mistakes
   │
Model 3
   │
Final Prediction
```

# Goal
Previous errors ko reduce karna

Strong model banana

# Example
✅ AdaBoost

✅ Gradient Boosting

✅ XGBoost

✅ LightGBM

✅ CatBoost

# 3. Stacking

👉 Multiple alag-alag models train hote hain.

Unki predictions ko ek Meta Model (Final Model) input ke roop me leta hai aur final prediction karta hai.

# Workflow

```
              Dataset
                 │
      ┌──────────┼──────────┐
      │          │          │
   Model 1    Model 2    Model 3
      │          │          │
      └──────────┼──────────┘
                 │
          Meta Model
                 │
        Final Prediction
```

# Goal
Har model ki strengths ka fayda uthana.
# Example
Decision Tree + SVM + Logistic Regression

In sabki predictions → Logistic Regression (Meta Model)

# Trick

```
Bagging = Baaju-baaju → Sab models saath-saath (parallel).
Boosting = Boost karo → Har naya model pichle ki mistake sudharta hai.
Stacking = Stack (layers) → Sab models ki predictions ko ek aur model combine karta hai.

# Why ensemble learning work

Pehle ek sawal 🤔

Maan lo tumhare class me 5 students hain.

Teacher ne ek difficult question pucha.

Answers:

Student 1 → Wrong

Student 2 → Correct

Student 3 → Correct

Student 4 → Wrong

Student 5 → Correct

Agar tum sirf Student 1 se puchte?

❌ Wrong answer milta.

Agar sabse puchte aur majority lete?

Correct = 3

Wrong = 2

Final = Correct

Accuracy badh gayi.

Yahi Ensemble Learning ka basic idea hai.

# Reason 1: Ek model perfect nahi hota

Har Machine Learning model ki kuch weaknesses hoti hain.

Example:

Model A

✓ Dog

✓ Cat

✗ Horse

Model B

✓ Horse

✗ Cat

✓ Dog

Model C

✓ Cat

✓ Horse

✗ Dog

Har model alag jagah mistake karta hai.

Jab in sabko combine karte ho,

to ek model ki galti ko doosra model cover kar deta hai.

Isliye final prediction better hoti hai.

# Reason 2: Errors cancel out

Maan lo prediction hai:

Model A → 90

Model B → 110

Model C → 100

Actual value hai

100

Notice karo

A ne -10 ki error ki.

B ne +10 ki error ki.

C bilkul sahi hai.

Average:

(90 + 110 + 100)/3

=100

Errors ek dusre ko balance kar gayi.

Isliye final prediction better bani.

# Reason 3: Har model alag perspective dekhta hai

Ek detective movie socho.

Ek witness bolta hai

Car red thi.

Dusra bolta hai

Driver black jacket me tha.

Teesra bolta hai

Car ka number yaad hai.

Teenon ke paas alag information hai.

Sabko combine karoge to complete picture milegi.

Models bhi exactly aisa hi karte hain.

# Reason 4: Noise ka effect kam ho jata hai

Agar ek model kisi noisy data ki wajah se confuse ho gaya,

to baaki models shayad confuse na hon.

Isliye final prediction stable hoti hai.



# Ensemble Learning works because every model has its own strengths and weaknesses. Different models usually make different mistakes. By combining the predictions of multiple models, these mistakes are reduced or balanced, resulting in a more accurate, stable, and reliable final prediction than a single model.

# When to Use

Jab single model se desired accuracy na mile ya prediction ko aur reliable banana ho, tab Ensemble Learning use karte hain.

# Use Ensemble Learning when a single model is not accurate enough, the dataset is complex, overfitting needs to be reduced, or high and stable prediction accuracy is required. Avoid it when you need simple, fast, and highly interpretable models or have limited computational resources.

# Voting Ensemble

Isme multiple models apni prediction dete hain, aur un sabki prediction ko vote karake final prediction nikali jaati hai.

Voting Ensemble ek technique hai jisme multiple machine learning models ki predictions ko combine karke voting ke basis par final prediction decide ki jaati hai.

#Example

Maan lo tumhare paas 3 models hain.

Question:

Ye image Cat hai ya Dog?

Model A → Cat

Model B → Dog

Model C → Cat

Ab votes count karte hain.

Cat = 2 Votes

Dog = 1 Vote

Final Prediction

Cat ✅

Yehi Voting Ensemble hai.

# Types of Voting Ensemble

1. Hard Voting (Majority Voting)

Isme har model sirf class label deta hai.

# Example:

Model 1 → Cat

Model 2 → Dog

Model 3 → Cat

Votes:

Cat = 2

Dog = 1

Final Answer:

Cat

👉 Jo class sabse zyada votes leti hai, wahi final prediction hoti hai.

# 2. Soft Voting (Probability Voting)

Isme models sirf class nahi batate, balki probability bhi dete hain.

# Example

```
| Model | Cat  | Dog  |
| ----- | ---- | ---- |
| A     | 0.90 | 0.10 |
| B     | 0.60 | 0.40 |
| C     | 0.80 | 0.20 |
```
Average probability:

Cat = (0.90 + 0.60 + 0.80) / 3 = 0.77

Dog = (0.10 + 0.40 + 0.20) / 3 = 0.23

Final Prediction:

Cat

👉 Soft Voting confidence (probabilities) ko bhi consider karta hai, isliye ye aksar Hard Voting se better perform karta hai.

# Advantages
✅ Simple to implement.

✅ Accuracy improve ho sakti hai.

✅ Ek model ki mistake ko doosre models balance kar dete hain.

✅ Stable predictions milti hain.

````python
from sklearn.datasets import load_iris
import pandas as pd

# Load the Iris dataset
iris = load_iris()
df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
df['target'] = iris.target
df['species'] = df['target'].map({0: 'setosa', 1: 'versicolor', 2: 'virginica'})

print(df.head())
````

````python
X = df.iloc[:,0:2]
y = df.iloc[:,-1]
````

````python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_val_score
````

````python
clf1 = LogisticRegression()
clf2 = RandomForestClassifier()
clf3 = KNeighborsClassifier()
````

````python
estimators = [('lr',clf1),('rf',clf2),('knn',clf3)]
````

````python
for estimator in estimators:
    x = cross_val_score(estimator[1],X,y,cv=10,scoring='accuracy')
    print(estimator[0],np.round(np.mean(x),2))
````

````python
from sklearn.ensemble import VotingClassifier
````

# Hard Voting

````python
vc = VotingClassifier(estimators=estimators,voting='hard')
x = cross_val_score(vc,X,y,cv=10,scoring='accuracy')
print(np.round(np.mean(x),2))
````

# Soft Voting

````python
vc1 = VotingClassifier(estimators=estimators,voting='soft')
x = cross_val_score(vc1,X,y,cv=10,scoring='accuracy')
print(np.round(np.mean(x),2))
````

# Weighted Voting

````python
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vc = VotingClassifier(estimators=estimators,voting='soft',weights=[i,j,k])
            x = cross_val_score(vc,X,y,cv=10,scoring='accuracy')
            print("for i={},j={},k={}".format(i,j,k),np.round(np.mean(x),2))
````

````python
from sklearn.svm import SVC
````

````python
from sklearn.datasets import make_classification
from sklearn.model_selection import cross_val_score
import numpy as np

X, y = make_classification(n_samples=1000, n_features=20, n_informative=15, n_redundant=5, random_state=2)

svm1 = SVC(probability=True, kernel='poly', degree=1)
svm2 = SVC(probability=True, kernel='poly', degree=2)
svm3 = SVC(probability=True, kernel='poly', degree=3)
svm4 = SVC(probability=True, kernel='poly', degree=4)
svm5 = SVC(probability=True, kernel='poly', degree=5)

estimation_estimators = [('svm1',svm1),('svm2',svm2),('svm3',svm3),('svm4',svm4),('svm5',svm5)]

for estimator in estimation_estimators:
    x = cross_val_score(estimator[1],X,y,cv=10,scoring='accuracy')
    print(estimator[0],np.round(np.mean(x),2))
````

````python
from sklearn.ensemble import VotingClassifier

vc1 = VotingClassifier(estimators=estimators,voting='soft')
x = cross_val_score(vc1,X,y,cv=10,scoring='accuracy')
print(np.round(np.mean(x),2))
````

# Code Voting_regressor

````python
from sklearn.datasets import fetch_california_housing
import numpy as np
````

````python
housing = fetch_california_housing()
X, y = housing.data, housing.target
````

````python


X.shape
````

````python

y.shape
````

````python
X
````

````python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
````

````python

lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
````

````python
estimators = [('lr',lr),('dt',dt),('svr',svr)]
````

````python

for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
````

````python

from sklearn.ensemble import VotingRegressor
````

````python

vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
````

````python
for i in range(1,4):
  for j in range(1,4):
    for k in range(1,4):
      vr = VotingRegressor(estimators,weights=[i,j,k])
      scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
      print("For i={},j={},k={}".format(i,j,k),np.round(np.mean(scores),2))
````

````python
# using the same algorithm

dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)
````

````python
estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
````

````python
for estimator in estimators:
  scores = cross_val_score(estimator[1],X,y,scoring='r2',cv=10)
  print(estimator[0],np.round(np.mean(scores),2))
````

````python

vr = VotingRegressor(estimators)
scores = cross_val_score(vr,X,y,scoring='r2',cv=10)
print("Voting Regressor",np.round(np.mean(scores),2))
````
