# Bagging ek technique hai jisme ek hi dataset se kai random datasets banaye jaate hain, har dataset par alag model train hota hai, aur sab models ki prediction ko combine karke final prediction diya jaata hai. Isse prediction zyada stable aur accurate ho jaati hai.

# Main Goal

✅ Ek model par depend na rehna

✅ Galti kam karna

✅ Prediction ko stable banana

✅ Overfitting kam karna (especially Decision Trees me)

```
| Model                         | Bias  | Variance | Reason                                                                                                             |
| ----------------------------- | ----- | -------- | ------------------------------------------------------------------------------------------------------------------ |
| Decision Tree                 | ✅ Low | ❌ High   | Data ko bahut closely fit karta hai, isliye data badalte hi tree badal jata hai.                                   |
| Bagging (many Decision Trees) | ✅ Low | ✅ Lower  | Bahut saare trees ki prediction combine hoti hai, isliye individual trees ki instability average out ho jaati hai. |
```

# "Bagging ka main purpose variance ko reduce karna hai. Isme multiple models ko alag-alag bootstrap samples par train kiya jata hai aur unki predictions ko average ya majority voting se combine kiya jata hai. Kyunki ek model ki galti dusre models balance kar dete hain, final prediction zyada stable hoti hai. Isliye Bagging variance ko kam karta hai, jabki bias me bahut kam ya lagbhag koi change nahi hota."

````python
# Define the evaluate function
# This function should ideally be defined in an earlier cell (e.g., with other imports),
# but for demonstration purposes to fix the NameError in this specific cell, it is defined here.
def evaluate(model, X_train, y_train, X_test_data, y_test_data):
    """
    Evaluates a given model on training and test data.
    Prints a warning if a classifier is used with continuous target variables.
    """
    model.fit(X_train, y_train)
    y_pred_train = model.predict(X_train)
    y_pred_test = model.predict(X_test_data)

    print(f"Model: {model.__class__.__name__}")
    print(f"Training Data Shape: {X_train.shape}, Target Shape: {y_train.shape}")
    print(f"Test Data Shape: {X_test_data.shape}, Target Shape: {y_test_data.shape}")

    # Check if the target is continuous or categorical (simple heuristic)
    if np.issubdtype(y_train.dtype, np.number) and len(np.unique(y_train)) > 10:
        print("\nWarning: The target variable appears to be continuous (e.g., 'median_house_value').")
        print("Using a DecisionTreeClassifier for a continuous target is generally incorrect.")
        print("For regression tasks, consider using DecisionTreeRegressor and appropriate regression metrics (e.g., R2 score, Mean Squared Error).")
        from sklearn.metrics import r2_score, mean_squared_error
        try:
            r2_train = r2_score(y_train, y_pred_train)
            mse_train = mean_squared_error(y_train, y_pred_train)
            r2_test = r2_score(y_test_data, y_pred_test)
            mse_test = mean_squared_error(y_test_data, y_pred_test)
            print(f"Training R2 Score: {r2_train:.4f}")
            print(f"Training MSE: {mse_train:.2f}")
            print(f"Test R2 Score: {r2_test:.4f}")
            print(f"Test MSE: {mse_test:.2f}")
        except Exception as e:
            print(f"Could not calculate regression metrics: {e}")
    else: # Assume classification if not clearly continuous
        from sklearn.metrics import accuracy_score
        try:
            train_accuracy = accuracy_score(y_train, y_pred_train)
            test_accuracy = accuracy_score(y_test_data, y_pred_test)
            print(f"Training Accuracy: {train_accuracy:.4f}")
            print(f"Test Accuracy: {test_accuracy:.4f}")
        except ValueError as e:
            print(f"Error calculating accuracy_score: {e}")
            print("This usually happens if target labels are not compatible (e.g., continuous values for classification metric).")

# Now call the function with the model, training data, and test data
evaluate(dt_bag1,X,y,X_test,y_test)
````

# Bagging Classifier

"Bagging Classifier ek machine learning algorithm hai jo ek classifier ki jagah bahut saare same classifiers ko alag-alag random data par train karta hai aur un sabki majority voting se final class predict karta hai."

# Example

Maan lo tumhare paas 5 Decision Trees hain.

Tree 1 → Cat

Tree 2 → Dog

Tree 3 → Dog

Tree 4 → Dog

Tree 5 → Cat

Majority vote:

Dog = 3 votes

Cat = 2 votes

✅ Final Prediction = Dog

# Bagging Regressor

Bagging Regressor ek machine learning algorithm hai jo regression problems (continuous values jaise house price, salary, temperature) ke liye bahut saare same regressors ko alag-alag bootstrap samples par train karta hai aur un sabki predictions ka average lekar final prediction deta hai."

# Example

House price predict karni hai.

5 models predict karte hain:

Model 1 → ₹48 Lakh

Model 2 → ₹50 Lakh

Model 3 → ₹49 Lakh

Model 4 → ₹51 Lakh

Model 5 → ₹52 Lakh

Final prediction:

(48 + 50 + 49 + 51 + 52) / 5
= ₹50 Lakh

Yaani average prediction final answer ban jaati hai.

# Bagging vs Voting

```
| **Bagging**                                                                                            | **Voting**                                                                              |
| ------------------------------------------------------------------------------------------------------ | --------------------------------------------------------------------------------------- |
| Ek hi dataset se **alag-alag bootstrap samples** (random sampling with replacement) banaye jaate hain. | Sab models **same original dataset** par train hote hain.                               |
| Har model ko **different bootstrap sample** milta hai.                                                 | Har model ko **same data** milta hai.                                                   |
| Aksar **same type ke models** hote hain (jaise 100 Decision Trees).                                    | Aksar **different type ke models** hote hain (Decision Tree, SVM, Logistic Regression). |
```

 # Example

 # Bagging

 Original Data:

1000 rows

Bootstrap Sampling:

Tree 1 → Random 1000 rows (duplicates ho sakte hain)

Tree 2 → Random 1000 rows

Tree 3 → Random 1000 rows

Tree 4 → Random 1000 rows

👉 Har tree different data par train hota hai.

# Goal: Variance kam karna

# Voting

Original Data:

1000 rows

Training:

Decision Tree → Same 1000 rows

SVM           → Same 1000 rows

Logistic Reg. → Same 1000 rows

KNN           → Same 1000 rows

👉 Sab models same data par train hote hain, bas algorithm alag hote hain.

# Goal: Different models ki strengths ko combine karna.
